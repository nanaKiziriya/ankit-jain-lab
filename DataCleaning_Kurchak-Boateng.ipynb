```python
#@title ## Data Format

#@markdown #### **[ Import Data ]**
#@markdown >##### Enter file path for data table:
#@markdown >##### E.g. */content/myData.xlsx* or */home/Jane/Desktop/myData.csv*
#@markdown >##### Must contain *time*, *variable* group(s), and *control* group. For current program, replicates must have _merged cell headings_.
dirty_path = '/content/AmK9_R9_SHEET-Raw-data_DROPg1gE-1t2.xlsx' #@param {type:'string'}

#@markdown <br>

#@markdown #### **[ Time ]**
#@markdown >##### Only decimal or H:M:S format accommodated. Must be ordered in 1st column.
#@markdown >##### Indicate if time is in H:M:S format. _(Will parse to seconds.)_
HMS = False #@param {type:'boolean'}
#@markdown >##### Enter unit of time:
unit_t = 'Minutes' #@param {type:'string'}
if HMS : unit_t = 'Seconds'

#@markdown <br>

#@markdown #### **[ Independent Variable ]**
#@markdown >##### Enter unit of independent variable:
unit_indep = 'Amyloid conc.' #@param {type:'string'}
#@markdown >##### Is there a control/blank group/trial?
has_ctrl = True #@param {type:'boolean'}
#@markdown >##### If so, if control/blank is NOT contained in the final column(s) then a _unique_ keyword indicating the relevant column header(s) must be provided. Make sure NO group names contain *'Unnamed'*.
target = 'Unnamed'
need_ctrl_kw = False #@param {type:'boolean'}
if need_ctrl_kw : ctrl_kw = '' #@param {type:'string'}

#@markdown <br>

#@markdown #### **[ Dependent Variable ]**
#@markdown >#####  Enter unit of measurement:
unit_y_dirty = 'Fluorescence' #@param {type:'string'}
```

```python
#@title ## Adv. Choices

#@markdown #### **[ Cut Control ]**
#@markdown >##### (If relevant) Subtract control from variable trials?
# DEPRECATED: (Note: If >1 replicates in a group later, *ctrl_sub* will become true.)
ctrl_sub = True #@param {type:'boolean'}
#@markdown >##### (If ctrl_sub TRUE) Subtract total average over trials _and_ over TIME? ... Else, is average over trials but _not_ over time.
ctrl_sub_avg = True #@param {type:'boolean'}
#@markdown >##### Discard control from final product?
ctrl_disc = True #@param {type:'boolean'}

if has_ctrl==False :
  ctrl_placeholder = 'CtrlPlaceholder'
  ctrl_sub = False
  ctrl_sub_avg = False
  ctrl_disc = True

#@markdown <br>

#@markdown #### **[ Truncation ]**
#@markdown >##### If time must be truncate _(e.g. systematic error)_ indicate so, and provide the number of datapoints to truncate on both ends. (Note: Any rows with NaN entries are dropped after truncation.)
truncate = True #@param {type:'boolean'}
if truncate : trunc_init = 3 #@param {type:'integer'}
if truncate : trunc_final = 0 #@param {type:'integer'}
if truncate : assert trunc_init>=0 and trunc_final>=0, 'trunc_init and trunc_final must be whole numbers'

#@markdown <br>

#@markdown #### **[ Alpha Normalization ]**
#@markdown >##### Normalize _alpha_ ==1 ? Else, original scale will be preserved.
normalize = False #@param {type:'boolean'}
#@markdown >##### If _alpha_ must be extrapolated _(e.g. equilibrium not reached)_ indicate so, and provide the model and extrapolated alpha.
extrapolated = False #@param {type:'boolean'}
if extrapolated : model = 'RK45' #@param {type:'string'}
alpha=0
if extrapolated : alpha = 0 #@param {type:'integer'}
if extrapolated : assert alpha>=0, 'alpha must be non-negative magnitude'
#@markdown >#####  Enter normalized unit of measurement:
unit_y_clean = 'Amyloid Conc.' #@param {type:'string'}

#@markdown <br>

#@markdown #### **[ Starting Point ]**
#@markdown >##### Theoretically & experimentally, does/should the head-end of the data line up?
#@markdown >##### _After all other cleaning, 1st datapoint will be scale equalized to their avg._
head_match = False #@param {type:'boolean'}

#@markdown <br>

#@markdown #### **[ Below-Blank Values ]**
#@markdown >##### If blanks are to be subtracted, after scaling, do you want resulting negative values to be zeroed?
neg_zero = True #@param {type:'boolean'}
```

```python
#@title ## Export format

#@markdown #### **[ Export Data ]**
#@markdown >##### Export cleaned data table?
exportData = True #@param {type:'boolean'}

#@markdown >##### Export plot images? (Note: *.jpeg* does not support transparency , *.svg* is vector-based)
exportImages = True #@param {type:'boolean'}
img_ext = '.png' #@param ['.png','.jpeg','.svg','pdf'] {type:'string'}
transparent = False #@param {type:'boolean'}


#@markdown #### <br>**[ Print program messages? (For debugging.) ]**
doPrint = True #@param {type:'boolean'}
```

```python
#@title _Useful Methods_
#@markdown

cond_print = lambda *s : print(*s) if doPrint else None

def groupAvgSeries(_pd,name=None) :
  if isinstance(_pd,pd.DataFrame) : # is df
    return pd.Series([
      np.average([
          entry
          for entry in
          _pd.iloc[row,:] ])
      for row in
      range(_pd.shape[0])
      ],name=name)
  else : # is series, not df
    if name : _pd.name=name
    return _pd



# PLOTTING

rgb = gitModule('nanaKiziriya','My_Computational_Tools','RGBSequence')
# rgb.color_sequence(num_colors=12, start=0, end=0,
#                exclude_end=True, reverse=False, ver='line',
#                alpha=255, channels=(0,1,2), chnl_mult=1, chnl_extreme=1, chnl_mix=0)
def rainbow(_N:int) :
  return rgb.color_sequence(
      num_colors=_N, end=9, exclude_end=False, chnl_mult=(1.3,.7,1.3))
ctrl_color = 'gray' # color for control group

def plotPandas(x_axis,y_pd,title:str=None,x_label:str=None,y_label:str=None,repl_:bool=False,color_=ctrl_color) :

  # series -> definitely no_repl_, single color provided, name is correct label
  if isinstance(y_pd,pd.Series) :
    plt.plot(x_axis, y_pd, color=color_, label=y_pd.name)

  # df -> color(s) provided
    # repl_ -> 1 color, 1 label [0]
    #       else -> colors [i], labels [i]
  elif isinstance(y_pd,pd.DataFrame) :
    _N = y_pd.shape[1]
    if not repl_ and color_==ctrl_color:
      color_ = rainbow(_N)
    for i in range(_N) :
      plt.plot(x_axis, y_pd.iloc[:,i],
               color=color_ if repl_ else color_[i],
               label=None if repl_ and i>0 else y_pd.columns[i])

  else :
    assert False, 'plotPandas not passed a series or df?'

  if x_label :
    plt.xlabel(x_label)
  if y_label :
    plt.ylabel(y_label)
  if title :
    plt.title=title



# FILE EXPORT

def _timestampPath(path) :
  dot = path.rfind('.')
  rtfn = dt.datetime.now()
  rtfn = f'_{rtfn.year}_{rtfn.month}_{rtfn.day}_{rtfn.hour}{rtfn.minute}{rtfn.second}'
  return path[:dot] + rtfn + path[dot:] if dot>-1 else path+rtfn

def _fixId(id:str) :
  return ('_' + id.strip(' _.')).replace(' ','_').replace('.','_')

def exportDataFrame(df:pd.DataFrame,id:str='ALTERED') :
  if not exportData: return

  dot = dirty_path.rfind('.')
  clean_path = dirty_path[:dot] + _fixId(id) + dirty_path[dot:]

  cleaner_path = _timestampPath(clean_path)
  while os.path.isfile(cleaner_path) :
    cleaner_path = _timestampPath(clean_path)

  try:
    df.to_excel(cleaner_path,index=False)
  except:
    print('Something went wrong while exporting',id,'data.')

def exportPlot(id='PYPLOT') :
  if not exportImages: return

  dot = dirty_path.rfind('.')
  clean_path = dirty_path[:dot] + _fixId(id) + img_ext

  cleaner_path = _timestampPath(clean_path)
  while os.path.isfile(cleaner_path) :
    cleaner_path = _timestampPath(clean_path)

  try:
    plt.savefig(cleaner_path, transparent=transparent)
  except:
    print(f'Something went wrong while exporting',id,'plot image.')
```

#**II. Preclean ⬇️**

```python
#@title ## Convert to Dataframe + Format
#@markdown

# Convert file to pandas dataframe
df_dirty = pd.read_excel(dirty_path,)
df_dirty.columns = df_dirty.columns.map(str) # avoid glitch when substr searching

if not has_ctrl :
  df_dirty[ctrl_placeholder] = 0

assert(df_dirty.shape[1]>2) # need 1 time column, at least 1 control, and at least 1 trial
# cond_print('Before truncation or dropping bottom NaN rows:\n', df_dirty)


# Drop NaN rows
i=0
n=0
nrows=df_dirty.shape[0]
while n<nrows :
  if('NaN' in str(df_dirty.iloc[i,:])) :
    df_dirty.drop(df_dirty.index[i], inplace=True)
  else : i+=1
  n+=1

# for _ in np.where(df_dirty == 'NaN')[0] :
#   df_dirty.drop(_, inplace=True)
# cond_print('\nAfter dropping NaN rows:\n', df_dirty)
```

```python
#@title ## Isolate time column
#@markdown

# BEFORE: df_dirty
# AFTER:  ser_time(formatted) | df_trials(no times)

ser_time = df_dirty.pop(df_dirty.columns[0])
assert(ser_time.name not in df_dirty.columns)

if HMS :
  temp = list()
  for i in range(len(ser_time)) :
    t = dt.datetime.strptime(str(ser_time[i]), "%H:%M:%S")
    temp.append(t.hour*3600+t.minute*60+t.second)
  ser_time = pd.Series(temp,name=ser_time.name)
cond_print(ser_time)

df_trials = df_dirty; del df_dirty # renaming
```

```python
#@title ### Identify Group Labels
#@markdown

# PURPOSE: When spreadsheet has merged column headings to represent replicate groups, pd.DataFrame uses label for 1st column in group, and names other group members 'Unnamed:##'

# target = 'Unnamed'

# Store actual group labels
  # there should be at least 2 valid labels (var and ctrl trial(s))
  # if there are less than 2 the program will throw a fit. :)
_group_labels = list(filter(
    lambda _:target not in _,
    df_trials.columns
))
assert len(_group_labels)>=2, f'Relabel columns in data table. Merge cells of labels for trials within the same replicate group. Do not use keyword \'{target}\'.'

# Identify control group label
_ctrl_label = (
    tuple(filter( lambda _:ctrl_kw in _, _group_labels))
    if need_ctrl_kw
    else _group_labels[-1])
if need_ctrl_kw :
  assert len(list(_ctrl_label))==1, f'need_ctrl_kw indicated, but ctrl_kw not unique: {ctrl_kw} in {_ctrl_label}'
  del ctrl_kw
```

```python
#@title ### Separate replicate groups
#@markdown

# BEFORE: ser_time | df_trials
# AFTER:  ser_time | dict_groups_pd(df_trials sep. by group)

# dict keys maintain order of insertion <3
# dict threadsafe: immutable during iteration -> pop from df_trials to new dict
dict_groups_pd = dict() # {group_N_label : df_group_N}

# procedure is same regardless of num replicates or ctrl column iloc

# Split df_trial into {label:panda} in dict
def next_label() :
  return df_trials.columns[0] if df_trials.shape[1]>0 else ''

for _lbl in _group_labels :
  _list = [df_trials.pop(next_label())]
  assert _list[0].name==_lbl, f'why {_list[0].name} no matchy {_lbl}? Maybe replicate trials column headers were not merged. For this program, please merge, or equivalently, delete headers of replcates after the 1st one.'

  while target in next_label() :
    _list.append(df_trials.pop(next_label()))

  # cond_print(_list)
  dict_groups_pd.update({_lbl : pd.concat(_list,axis=1)})

# dict keys are correct group names, dict vals (df) column labels prolly fked

assert(df_trials.shape[1]==0)
del df_trials

assert all(_ in dict_groups_pd.keys() for _ in _group_labels)
# assert _group_labels[_]==dict_groups_pd.keys()[_] for _ in range(len(_group_labels))
del _group_labels, target

if doPrint:
  cond_print(f'After split:')
  for group, pd_ in dict_groups_pd.items() :
    cond_print(group,pd_,'\n')
```

```python
#@title ### Separate the control group
#@markdown

# BEFORE: ser_time | dict_groups_pd
# AFTER:  ser_time | dict_vgroups_pd(variable groups only, no ctrls) | pd_ctrls

pd_ctrls = dict_groups_pd.pop(_ctrl_label)

cond_print(f'CONTROLS:\n{pd_ctrls}')

dict_vgroups_pd = dict_groups_pd; del dict_groups_pd # rename
```

```python
#@title ### **INITIAL VISUALIZATION:**
#@markdown
# Note: data was not altered yet, only moved around

_color = rainbow(len(dict_vgroups_pd))

for _, (_grp,_pd) in enumerate(dict_vgroups_pd.items()):
  plotPandas(ser_time, _pd, repl_=True, color_=_color[_], title='Raw Trial Data', x_label=unit_t, y_label=unit_y_dirty)

if has_ctrl :
  plotPandas(
    ser_time,
    pd_ctrls,
    repl_=True)

plt.legend(title=unit_indep)
exportPlot('DIRTYPLOT')
plt.show()
```

```python
#@title ## *Truncation*
#@markdown

# Head-tail-end truncation
if truncate :
  __ = ser_time.size
  # for _ in [ser_time,pd_ctrls]+list(dict_vgroups_pd.values()) :
  #   _ = _.iloc[trunc_init : __-1*trunc_final]
  _newrange = range(ser_time.size-trunc_init-trunc_final)
  ser_time = ser_time.iloc[trunc_init : __-1*trunc_final]
  ser_time.index=_newrange
  pd_ctrls = pd_ctrls.iloc[trunc_init : __-1*trunc_final]
  pd_ctrls.index = _newrange
  for _ in dict_vgroups_pd.keys() :
    dict_vgroups_pd[_] = dict_vgroups_pd[_][trunc_init : __-1*trunc_final]
    dict_vgroups_pd[_].index=_newrange


  cond_print('After truncation:\n', ser_time)
  cond_print(list(dict_vgroups_pd.values())[0])
  cond_print('...')
  cond_print(pd_ctrls)
```

```python
#@title #**1. Averaging Control Trials**

ser_ctrl = groupAvgSeries(pd_ctrls,name=_ctrl_label) if has_ctrl else pd_ctrls

if ctrl_sub_avg :
  ser_ctrl = pd.Series([ser_ctrl.mean()]*ser_ctrl.size,name=ser_ctrl.name)

del pd_ctrls, _ctrl_label
# Now we have: ser_time | dict_vgroups_pd | ser_ctrl
```

```python
#@title #**2. Subtract Control Baseline From All Other Trials**

replicates = not all(_.size == ser_time.size for _ in dict_vgroups_pd.values())
# ctrl_sub = ctrl_sub or replicates

if ctrl_sub :
  cond_print(f'Before subtracting ctrl baseline:\n{dict_vgroups_pd}\n')
  for group in dict_vgroups_pd :
    dict_vgroups_pd[group] = dict_vgroups_pd[group].sub(ser_ctrl, axis=0)
  if not ctrl_disc : ser_ctrl = pd.Series([0]*ser_ctrl.size,name=ser_ctrl.name)
  cond_print(f'After subtracting ctrl baseline:\n{dict_vgroups_pd}\n')
```

```python
#@title #**3. Prescale trials within each group**
# Within in each replicate group, make sure the average of all data points match up for each trial
# scale UP first to avoid precision loss from truncation

# make sure scaling up does not create overflow error
# to be safe, largest value to power of len_prescale should not overflow

# import sys
# MAX = pd.concat(dict_vgroups_pd.values()).abs().max()
# try :
#   MAX = max(MAX.abs())
# except :
#   cond_print(f'oops {MAX} is already the max :)')
# assert(MAX < np.power(sys.float_info.max, (1/len_prescale)))
# # IF IT CREATES OVERFLOW, THE FOLLOWING ALGORITHM HAS TO BE REPROGRAMMED. I AM NOT DOING ALL THAT RIGHT NOW.

for group,df in dict_vgroups_pd.items() :
  if isinstance(df,pd.DataFrame) :
    prescale_avgs = list(map(
        lambda _:np.average(df.iloc[:,_]), range(df.shape[1])
        ))
    total_avg = np.average(prescale_avgs)
    prescale_factors = list(map(lambda _:total_avg/_,prescale_avgs))

    # multiply each column j by prescale_factors[j] all at once using df.mul
    assert(len(prescale_factors)==(df.shape[1]))
    dict_vgroups_pd[group] = df.mul(prescale_factors, axis=1)

cond_print(f'After prescaling:\n',dict_vgroups_pd)

# Note: prescaling is to balance statistical weight of each trial before averaging them
```

```python
#@title #**4. Average trials in each group; neg-zero and head-match if relevant**
# Then put all group columns into a single dataframe

# avg within each group (scale preserved)
df_vgroups = pd.concat((groupAvgSeries(df,name=group) for group,df in dict_vgroups_pd.items()), axis=1)

if neg_zero :
  df_vgroups[df_vgroups < 0] = 0

if head_match :
  prescale_avgs = df_vgroups.iloc[0,:]
  total_avg = prescale_avgs.mean()
  prescale_factors = list(map(lambda _:total_avg/_,prescale_avgs))
  df_vgroups = df_vgroups.mul(prescale_factors, axis=1)

cond_print(df_vgroups)
```

```python
#@title #**5. Normalize entire set of values: Scale onto [0,1]**
# max value in df_vgroups should =1

if not extrapolated : alpha = df_vgroups.max().max()

if normalize :
  df_vgroups = df_vgroups.div(alpha)
  if not (ctrl_disc or ctrl_sub) : ser_ctrl.div(alpha)

  if not extrapolated : assert(1==df_vgroups.max().max())

df_vgroups
```

#**In Conclusion...**

```python
#@title ### **FINAL VISUALIZATION:**
_color = list(rainbow(df_vgroups.shape[1]))


if not ctrl_disc :
  _color.append(ctrl_color)
  df_vgroups = pd.concat([df_vgroups, ser_ctrl],axis=1)

plotPandas(
    ser_time,
    df_vgroups,
    title='Normalized Data',
    x_label=unit_t,
    y_label=unit_y_clean,
    repl_=False,
    color_=_color)

plt.legend(title=unit_indep)

exportPlot(id='CLEANPLOT')
plt.show()
```
